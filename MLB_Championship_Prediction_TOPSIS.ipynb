```python
#Korelasyonlar arası nitelikli analiz içi z üzeriden p değeri çıkarımı(OPS Kıyasında aktif)
import numpy as np
from scipy.stats import norm

def fisher_z_test(r1, r2, n1, n2):

    z1 = 0.5 * np.log((1 + r1) / (1 - r1))
    z2 = 0.5 * np.log((1 + r2) / (1 - r2))

    se = np.sqrt(1/(n1-3) + 1/(n2-3))

    z_score = (z1 - z2) / se

    p_value = 2 * (1 - norm.cdf(abs(z_score)))

    return z_score, p_value

r1, r2 = 0.620, 0.813
n1, n2 = 10, 10

z_score, p_value = fisher_z_test(r1, r2, n1, n2)
print(f"Z-score: {z_score}, P-value: {p_value}")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

data = {
    "BatterWAR": [20.8, 33.21, 21.1, 26.3, 41],
    "PitcherWAR": [17.7, 22.41, 19.5, 22.9, 14.1],
    "BestPlayerBatterWar": [5.3, 8.9, 6.3, 6.4, 8.4],
    "BestPlayerPitcherWar": [6.0, 7.8, 4.2, 6.0, 3.0],
    "Şampiyon Oldu?": [0.0001, 0.0001, 1, 0.0001, 0.0001]
}
df = pd.DataFrame(data)

X = df[["BatterWAR",'PitcherWAR' ,"BestPlayerBatterWar", "BestPlayerPitcherWar"]]
y = df["Şampiyon Oldu?"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Model Doğruluk Oranı: {accuracy:.2f}")
print("Sınıflandırma Raporu:")
print(classification_report(y_test, y_pred))

new_customer = np.array([[32.24, 26.83, 7.04, 5.61]]) #2025 tahminleri
prediction = model.predict(new_customer)
probabilities = model.predict_proba(new_customer)

print(f"Yeni sezon tahmini (1=Ciddi bir şampiyonluk adayı, 0=Zor gibi): {prediction[0]}")
if probabilities.shape[1] == 2:
    print(f"Tahmin olasılıkları: {probabilities[0][1]*100:.2f}% '1' (Şampiyonluk) ve {probabilities[0][0]*100:.2f}% '0' (Şampiyonluk değil)")
else:
    print(f"Tahmin olasılığı: {probabilities[0][0]*100:.2f}%")
```
